```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/jobs.csv")
df.head()
```

```python
df = df.drop("Unnamed: 0", axis=1)
```

```python
df["Founded"] = pd.to_numeric(df["Founded"], errors="coerce")
```

```python
df["Size"] = df["Size"].str.replace(" employees", "")
```

```python
df["Revenue"] = df["Revenue"].str.replace(" USD", "")
```

```python
df["Easy Apply"] = df["Easy Apply"].notnull().astype(int)
```

```python
df.isnull().sum()
```

```python
skills_list = [
    "python", "sql", "excel", "power bi", "tableau",
    "machine learning", "deep learning", "nlp",
    "aws", "azure", "gcp",
    "pandas", "numpy", "matplotlib", "seaborn"
]
```

```python
def extract_skills(text):
    text = str(text).lower()
    found_skills = []
    
    for skill in skills_list:
        if skill in text:
            found_skills.append(skill)
    
    return found_skills
```

```python
df["Skills"] = df["Job Description"].apply(extract_skills)
```

```python
df[["Job Title", "Skills"]].head()
```

```python
df_exploded = df.explode("Skills")
```

```python
top_skills = df_exploded["Skills"].value_counts()
print(top_skills.head(10))
```

```python
df.to_csv("../data/processed/cleaned_jobs.csv", index=False)
df_exploded.to_csv("../data/processed/skills_data.csv", index=False)
```
